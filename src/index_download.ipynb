```python
from pathlib import Path
import akshare as ak
import pandas as pd

df = ak.index_zh_daily_tx(symbol="sh000300")

if df.empty:
    raise ValueError("接口未返回指定日期范围的数据，请检查返回内容。")

# 插入etf的编号
df.insert(0, "index_code", "000300")

print(df.head())

print(df.head())
print(df.tail())
print("数据规模:", df.shape)
print("日期范围:", df["date"].min(), "至", df["date"].max())

# 使用明确的项目路径，避免 notebook 工作目录影响保存位置。
output_path = Path(r"D:\Project\Quant-ETF\data\raw\000300_index.csv")
output_path.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(output_path, index=False, encoding="utf-8-sig")
print("已保存:", output_path)
```
