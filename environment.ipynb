```python
# 新建环境位置
conda create --prefix "D:\CondaData\envs\quant_env" --override-channels -c https://mirrors.tuna.tsinghua.edu.cn/anaconda/cloud/conda-forge python=3.12 pip -y
conda activate "D:\CondaData\envs\quant_env"

python --version
python -m pip --version
python -c "import sys; print(sys.executable)"
```

```python
# 安装基本库
$pipIndex = "https://pypi.org/simple"
python -m pip install --dry-run --only-binary=:all: --index-url $pipIndex -r .\quant_env.txt --report .\core_install_plan.json
python -m pip install --only-binary=:all: --index-url $pipIndex -r .\quant_env.txt
python -m pip check
python -c "import numpy, scipy, pandas, matplotlib; print('Core packages OK')"
```

```python
# 安装特殊库
python -m pip install --only-binary=:all: --index-url $pipIndex -c akshare openpyxl pyarrow
python -m pip check
```

```python
# 配置jupyter notebook
python -m pip check
python -m ipykernel install --user --name quant_env --display-name "Python (quant_env)"
python -c "import sys, ipykernel; print('Python:', sys.executable); print('ipykernel:', ipykernel.__file__)"
```

```python
# 保存已经安装的库的信息
python -m pip freeze | Out-File -Encoding utf8 requirements.txt
```

```python
# 下次在其他电脑上安装
python -m pip install -r requirements.txt # 只有从清单中批量下载才用-r
```
